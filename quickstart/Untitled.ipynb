```python
import pandas as pd
import os
from pathlib import Path

def create_csv_files(df, output_dir, num_frames=1600):
    """
    Create separate CSV files for each row in the DataFrame.
    
    Args:
        df: pandas DataFrame with columns including 'Full File Name', 'param1', 'param0'
        output_dir: path where CSV files should be written
    """
    # Create output directory if it doesn't exist
    os.makedirs(output_dir, exist_ok=True)
    
    for idx, row in df.iterrows():
        # Extract filename without extension
        full_filename = row['Full File Name']
        filename_no_ext = Path(full_filename).stem
        csv_filename = f"{filename_no_ext}.csv"
        csv_path = os.path.join(output_dir, csv_filename)
        
        # Create DataFrame with 16000 rows
        csv_df = pd.DataFrame({
            'param1': [row['param1']] * num_frames,
            'class': [row['param0']] * num_frames
        })
        
        # Write to CSV
        csv_df.to_csv(csv_path, index=False)
        
        if (idx + 1) % 100 == 0:
            print(f"Processed {idx + 1} files...")
    
    print(f"Done! Created {len(df)} CSV files in {output_dir}")

# Usage:
# create_csv_files(table.to_pandas(), '/path/to/output/directory')
```
