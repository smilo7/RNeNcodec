```python
# QuickStart: 2_Inference (offline + real-time demo)
%load_ext autoreload
%autoreload 2
%pip install -e .
%pip install -e ./synth[ui]

import subprocess, yaml
from pathlib import Path

# Ensure weights exist (auto-download if missing)
cfg = yaml.safe_load(open("rnencodec/configs/quickstart_infer.yaml", "r"))
ckpt = Path(cfg["checkpoint"])
if not ckpt.exists():
    print("Downloading pretrained weights...")
    subprocess.run(["python", "scripts/download_artifacts.py", "--weights"], check=True)

# Instantiate generator (adapt to your API)
from rnencodec.generator.generator import RNNGenerator
gen = RNNGenerator.from_checkpoint(ckpt, device=cfg["device"])  # implement @classmethod
gen.prime(warmup_frames=cfg["warmup_frames"])

# Offline generate 5s and preview
audio = gen.generate(seconds=5.0, sample_rate=cfg["sample_rate"])  # ndarray
from IPython.display import Audio
Audio(audio, rate=cfg["sample_rate"])
```
